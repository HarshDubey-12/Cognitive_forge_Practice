```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)

rate = 3
inter_arrivals = np.random.exponential(1/rate,1000)
arrival_times = np.cumsum(inter_arrivals)
```

```python
plt.step(arrival_times, range(len(arrival_times)))
plt.title("Poisson Process Simulation")
plt.show()
```
